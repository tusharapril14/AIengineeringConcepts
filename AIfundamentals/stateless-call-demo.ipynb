```python
def stateless_api_call(prompt:str)->str:
    from openai import OpenAI
    from dotenv import load_dotenv
    load_dotenv()  # Load environment variables from .env file
    client = OpenAI()
    # Simulate an API call
    response = client.responses.create(
        model="gpt-4o-mini",
        input=prompt
    )

    return response.output_text
```

```python
stateless_api_call("Who is the president of united states?")
```

```python
stateless_api_call("Who was before him?")
```
