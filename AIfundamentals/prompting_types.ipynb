# Prompting types

This notebook compares **zero-shot**, **one-shot**, and **few-shot** prompting. Each example calls the same Groq helper, so the difference is only the context supplied to the model.

```python
from os import getenv

from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()


def groq_completion(messages: list[dict[str, str]]) -> str:
    api_key = getenv("GROQ_API_KEY")
    if not api_key:
        raise ValueError("Set GROQ_API_KEY in your environment or .env file.")

    client = OpenAI(
        api_key=api_key,
        base_url="https://api.groq.com/openai/v1",
    )
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=messages,
        temperature=0,
    )
    return response.choices[0].message.content
```

## Zero-shot prompting

A zero-shot prompt gives the model instructions and the task, with no examples of the expected answer. Use it when the task is clear and familiar to the model.

```python
zero_shot_messages = [
    {
        "role": "system",
        "content": "Classify the customer message as billing, technical_support, or account. Reply with only the label.",
    },
    {
        "role": "user",
        "content": "My card was charged twice for the same subscription.",
    },
]

print(groq_completion(zero_shot_messages))
```

## One-shot prompting

A one-shot prompt adds one input-output example. The example establishes the response format and demonstrates how to handle the task.

```python
one_shot_messages = [
    {
        "role": "system",
        "content": "Classify the customer message as billing, technical_support, or account. Reply with only the label.",
    },
    {"role": "user", "content": "I cannot reset my password."},
    {"role": "assistant", "content": "account"},
    {"role": "user", "content": "The mobile app closes as soon as I open it."},
]

print(groq_completion(one_shot_messages))
```

## Few-shot prompting

A few-shot prompt provides several representative examples. It is useful when labels are nuanced or the output format needs to be very consistent.

```python
few_shot_messages = [
    {
        "role": "system",
        "content": "Classify the customer message as billing, technical_support, or account. Reply with only the label.",
    },
    {"role": "user", "content": "Where can I download my invoice?"},
    {"role": "assistant", "content": "billing"},
    {"role": "user", "content": "Two-factor authentication is not sending me a code."},
    {"role": "assistant", "content": "account"},
    {"role": "user", "content": "The dashboard shows a blank page after I sign in."},
    {"role": "assistant", "content": "technical_support"},
    {"role": "user", "content": "I was billed after I cancelled my plan."},
]

print(groq_completion(few_shot_messages))
```
